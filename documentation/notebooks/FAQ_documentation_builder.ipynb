# Alma Analytics Catalog Parser FAQ Documentation Builder

This notebook organizes presentation and FAQ questions about the Alma Analytics catalog parser. Each entry contains a stable identifier, a category, a question, and the points an effective answer should cover.

## Project purpose

### Q: What is this project, what problem does it solve, and what does it produce?

- It is an Alma Analytics report-criteria exporter.
- It addresses the lack of a reusable export of underlying Analytics criteria and reporting logic.
- It parses Alma Analytics `.catalog` files.
- It extracts filters, saved columns, bins, criteria, and related metadata.
- It produces reusable CSV and Excel review outputs for human review
