### Project overview

- What problem does the parser solve?
    - It is an Alma Analytics report-criteria exporter.
    - It addresses the lack of a reusable export of underlying Analytics criteria and reporting logic.
    - It parses Alma Analytics `.catalog` files.
    - It extracts filters, saved columns, bins, criteria, and related metadata.
    - It produces reusable CSV and Excel review outputs for human review

- How is this different from exporting report results from Alma Analytics?
    - A results export contains the rows returned when an analysis runs. 
    - This project documents how Analytics objects are configured and why records are included or excluded.

- Who is the intended audience?
- What decisions become more transparent because of this project?

### Input and processing

- What is an Alma Analytics `.catalog` file?
    An Alma Analytics `.catalog` file is an exported package of objects from the Alma Analytics catalog.

    It can contain definitions for objects such as:

    - Analyses and reports
    - Saved filters
    - Saved columns
    - Bin definitions
    - Dashboards and dashboard pages
    - Object names, paths, signatures, and modification metadata

    The file is a compressed container containing catalog metadata and embedded XML—not a conventional spreadsheet or plain XML file. The project’s R parser decompresses it, separates the embedded XML objects, matches them with their catalog metadata, identifies each object type, and translates reporting criteria into reusable CSV and Excel tables.

    A `.catalog` file primarily describes how Analytics objects are configured. It is different from exporting a report’s resulting rows of data.
