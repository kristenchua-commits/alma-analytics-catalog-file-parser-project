# XML and object structure of a `.catalog` file

An Alma Analytics `.catalog` file is a compressed archive containing catalog metadata and many embedded XML documents—not one continuous XML document. This notebook summarizes the XML structure found in `data/examples/annual_stats_fy_2025_2026.catalog`. Repeated sibling indexes are combined so the tree emphasizes structure rather than displaying all 163,097 extracted tag occurrences.

| Analysis/report | 98 | `saw:report` |
| Dashboard | 22 | `sawd:dashboard` |
| Dashboard page | 103 | `sawd:dashboardPage` |
| **Total** | **308** | — |

## Regenerate the complete tree

Run the following R cell after running the main parser. It reads `output/xml_tag_inventory.csv`, removes numeric sibling indexes such as `[2]`, builds a hierarchy of all distinct tag paths, and displays a complete collapsible tree. This keeps the visualization synchronized with a newer `.catalog` export.

```r
inventory_path <- "output/xml_tag_inventory.csv"
catalog_path <- "output/catalog_extract.rds"

stopifnot(file.exists(inventory_path), file.exists(catalog_path))

inventory <- read.csv(inventory_path, stringsAsFactors = FALSE)
catalog <- readRDS(catalog_path)
inventory$normalized_path <- gsub("[[][^]]+[]]", "", inventory$tag_path)

escape_html <- function(value) {
  value <- gsub("&", "&amp;", value, fixed = TRUE)
  value <- gsub("<", "&lt;", value, fixed = TRUE)
  gsub(">", "&gt;", value, fixed = TRUE)
}

render_nodes <- function(path_parts, depth = 1L) {
  available <- lengths(path_parts) >= depth
  path_parts <- path_parts[available]
  if (!length(path_parts)) return("")

  node_names <- sort(unique(vapply(path_parts, `[[`, character(1), depth)))
  rendered <- vapply(node_names, function(node_name) {
    selected <- vapply(
      path_parts,
      function(parts) length(parts) >= depth && identical(parts[[depth]], node_name),
      logical(1)
    )
    matching_paths <- path_parts[selected]
    has_children <- any(lengths(matching_paths) > depth)
    label <- paste0("<code>", escape_html(node_name), "</code>")

    if (!has_children) {
      return(paste0("<div class='xml-leaf'>", label, "</div>"))
    }

    paste0(
      "<details><summary>", label, "</summary>",
      render_nodes(matching_paths, depth + 1L),
      "</details>"
    )
  }, character(1))

  paste(rendered, collapse = "")
}

object_kinds <- sort(unique(inventory$object_kind))
kind_trees <- vapply(object_kinds, function(kind) {
  kind_paths <- sort(unique(inventory$normalized_path[inventory$object_kind == kind]))
  path_parts <- strsplit(sub("^/", "", kind_paths), "/", fixed = TRUE)
  object_count <- sum(catalog$object_kind == kind)
  paste0(
    "<details><summary><strong>", escape_html(kind), "</strong> — ",
    object_count, " objects</summary>",
    render_nodes(path_parts),
    "</details>"
  )
}, character(1))

tree_html <- paste0(
  "<style>",
  "#generated-catalog-tree details{margin:.28rem 0 .28rem 1.1rem}",
  "#generated-catalog-tree>details{margin-left:0}",
  "#generated-catalog-tree summary{cursor:pointer;list-style:none}",
  "#generated-catalog-tree summary::-webkit-details-marker{display:none}",
  "#generated-catalog-tree summary::before{content:'+ ';font-weight:700}",
  "#generated-catalog-tree details[open]>summary::before{content:'− '}",
  "#generated-catalog-tree .xml-leaf{margin:.25rem 0 .25rem 2.35rem}",
  "</style>",
  "<div id='generated-catalog-tree'>",
  "<details open><summary><strong>Alma Analytics .catalog archive</strong> — ",
  nrow(catalog), " embedded XML objects</summary>",
  paste(kind_trees, collapse = ""),
  "</details></div>"
)

if (requireNamespace("IRdisplay", quietly = TRUE)) {
  IRdisplay::display_html(tree_html)
} else {
  cat(tree_html)
}
```
